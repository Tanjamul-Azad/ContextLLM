# 09 WP0.4 Backbone competence screen (development items, exploratory)

Runs one candidate backbone on the 20 no-example items (10 v1 draft cards + 10 pilot seeds) and the 68 oracle variants, scores them, and rebuilds the summary table with 95% bootstrap intervals. A model is eligible as a method backbone if it solves at least 80% of the no-example tasks. Greedy decoding, max 1024 new tokens; resumable. Code answers are executed by the checker, which is not sandboxed. Pre-quantised 4-bit checkpoints.
